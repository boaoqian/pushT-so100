```python
import numpy as np
from lerobot.datasets.lerobot_dataset import LeRobotDataset
import matplotlib.pyplot as plt
from env_gym import PushT
import cv2
from helper import get_test_image_transform

repo_id = "/media/qba/Data/Project/Robot/So100PushT/data/pusht_so100_dataset_merge"

dataset = LeRobotDataset(repo_id,episodes=[0])
print(dataset)
t = get_test_image_transform()
```

```python
print(dataset.meta.stats["action"])
mean = dataset.meta.stats["action"]["mean"]
std = dataset.meta.stats["action"]["std"]
```

```python
env = PushT("/media/qba/Data/Project/Robot/So100PushT/chernyadev mujoco_menagerie add-so-arm100 trs_so_arm100/test_env.xml")
env.reset()
import time
for d in dataset:
  a = d["action"]#*std+mean
  cam, _,_,_,_ = env.step(a)
  img_side = t(cam["cam_side"]).permute(1,2,0).numpy()
  img_top = t(cam["cam_top"]).permute(1,2,0).numpy()
  img_side = cv2.cvtColor(img_side, cv2.COLOR_RGB2BGR)
  img_top = cv2.cvtColor(img_top, cv2.COLOR_RGB2BGR)
  cv2.imshow("side", img_side)
  cv2.imshow("top", img_top)

  time.sleep(1/30)
  cv2.waitKey(1)
cv2.DistroyAllWindows()
```
